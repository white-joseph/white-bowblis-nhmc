```python
from pathlib import Path
import pandas as pd
import numpy as np
import re

# ------------------------------------------------------------
# Load data
# ------------------------------------------------------------
panel_fp = Path(r"C:/Repositories/white-bowblis-nhmc/data/interim/quality_measures_quarter_panel.csv")
out_dir = Path(r"C:/Repositories/white-bowblis-nhmc/data/interim")
out_dir.mkdir(parents=True, exist_ok=True)

qm = pd.read_csv(panel_fp, low_memory=False)

# ------------------------------------------------------------
# Identify quarter columns
# ------------------------------------------------------------
quarter_cols = [c for c in qm.columns if re.fullmatch(r"Q[1-4]_\d{4}", c)]

id_cols = [
    "cms_certification_number",
    "quality_metric_code",
    "quality_metric_description"
]

# ------------------------------------------------------------
# 1. Basic dimensions
# ------------------------------------------------------------
print("\n---------------- BASIC DIMENSIONS ----------------")
print(f"Rows: {len(qm):,}")
print(f"Columns: {qm.shape[1]:,}")
print(f"Quarter columns: {len(quarter_cols):,}")

# ------------------------------------------------------------
# 2. Missingness by column
# ------------------------------------------------------------
missing_summary = pd.DataFrame({
    "column": qm.columns,
    "n_missing": qm.isna().sum().values,
    "pct_missing": (qm.isna().mean().values * 100)
}).sort_values(["pct_missing", "n_missing"], ascending=[False, False])

print("\n---------------- MISSINGNESS BY COLUMN ----------------")
print(missing_summary.to_string(index=False))

missing_summary.to_csv(out_dir / "quality_measures_missing_by_column.csv", index=False)

# ------------------------------------------------------------
# 3. Duplicate row checks
# ------------------------------------------------------------
print("\n---------------- DUPLICATE CHECKS ----------------")

n_full_dupes = int(qm.duplicated().sum())
print(f"Full duplicated rows: {n_full_dupes:,}")

n_id_dupes = int(
    qm.groupby(id_cols, dropna=False)
      .size()
      .reset_index(name="n")
      .query("n > 1")
      .shape[0]
)
print(f"Duplicate rows by CCN + code + description: {n_id_dupes:,}")

n_code_dupes = int(
    qm.groupby(["cms_certification_number", "quality_metric_code"], dropna=False)
      .size()
      .reset_index(name="n")
      .query("n > 1")
      .shape[0]
)
print(f"Duplicate rows by CCN + code: {n_code_dupes:,}")

# ------------------------------------------------------------
# 4. Code-description consistency
# ------------------------------------------------------------
code_desc_pairs = qm[["quality_metric_code", "quality_metric_description"]].drop_duplicates()

code_desc_check = (
    code_desc_pairs.groupby("quality_metric_code", dropna=False)
    .size()
    .reset_index(name="n_descriptions")
    .sort_values(["n_descriptions", "quality_metric_code"], ascending=[False, True])
)

print("\n---------------- CODE -> DESCRIPTION CONSISTENCY ----------------")
print(code_desc_check.to_string(index=False))

code_desc_conflicts = (
    code_desc_pairs.merge(code_desc_check, on="quality_metric_code", how="left")
    .query("n_descriptions > 1")
    .sort_values(["quality_metric_code", "quality_metric_description"])
)

print("\nCodes with multiple descriptions:")
if len(code_desc_conflicts) == 0:
    print("None")
else:
    print(code_desc_conflicts.to_string(index=False))

code_desc_conflicts.to_csv(
    out_dir / "quality_measures_code_description_conflicts_from_panel.csv",
    index=False
)

# ------------------------------------------------------------
# 5. Quarter-level completeness
# ------------------------------------------------------------
quarter_completeness = pd.DataFrame({
    "quarter": quarter_cols,
    "n_nonmissing": qm[quarter_cols].notna().sum().values,
    "n_missing": qm[quarter_cols].isna().sum().values,
    "pct_nonmissing": qm[quarter_cols].notna().mean().values * 100,
    "pct_missing": qm[quarter_cols].isna().mean().values * 100
}).sort_values("quarter")

print("\n---------------- QUARTER COMPLETENESS ----------------")
print(quarter_completeness.to_string(index=False))

quarter_completeness.to_csv(
    out_dir / "quality_measures_quarter_completeness.csv",
    index=False
)

# ------------------------------------------------------------
# 6. Metric-level completeness
# ------------------------------------------------------------
qm_metric = qm.copy()
qm_metric["nonmissing_quarters"] = qm_metric[quarter_cols].notna().sum(axis=1)

metric_rows = []
for (code, desc), g in qm_metric.groupby(
    ["quality_metric_code", "quality_metric_description"], dropna=False
):
    quarter_array = g[quarter_cols].to_numpy()
    pct_cells_nonmissing = np.mean(~pd.isna(quarter_array)) * 100

    metric_rows.append({
        "quality_metric_code": code,
        "quality_metric_description": desc,
        "n_facilities": len(g),
        "avg_nonmissing_quarters": g["nonmissing_quarters"].mean(),
        "min_nonmissing_quarters": g["nonmissing_quarters"].min(),
        "max_nonmissing_quarters": g["nonmissing_quarters"].max(),
        "pct_cells_nonmissing": pct_cells_nonmissing
    })

metric_completeness = pd.DataFrame(metric_rows).sort_values(
    ["pct_cells_nonmissing", "quality_metric_code"],
    ascending=[False, True]
)

print("\n---------------- METRIC COMPLETENESS ----------------")
print(metric_completeness.to_string(index=False))

metric_completeness.to_csv(
    out_dir / "quality_measures_metric_completeness.csv",
    index=False
)

# ------------------------------------------------------------
# 7. Facility-level completeness
# ------------------------------------------------------------
facility_rows = []
for ccn, g in qm_metric.groupby("cms_certification_number", dropna=False):
    quarter_array = g[quarter_cols].to_numpy()
    pct_cells_nonmissing = np.mean(~pd.isna(quarter_array)) * 100

    facility_rows.append({
        "cms_certification_number": ccn,
        "n_metrics": len(g),
        "avg_nonmissing_quarters": g["nonmissing_quarters"].mean(),
        "min_nonmissing_quarters": g["nonmissing_quarters"].min(),
        "max_nonmissing_quarters": g["nonmissing_quarters"].max(),
        "pct_cells_nonmissing": pct_cells_nonmissing
    })

facility_completeness = pd.DataFrame(facility_rows).sort_values(
    ["pct_cells_nonmissing", "cms_certification_number"],
    ascending=[False, True]
)

print("\n---------------- FACILITY COMPLETENESS ----------------")
print(facility_completeness.head(50).to_string(index=False))

facility_completeness.to_csv(
    out_dir / "quality_measures_facility_completeness.csv",
    index=False
)

# ------------------------------------------------------------
# 8. All-quarters-missing rows
# ------------------------------------------------------------
all_quarters_missing = int(qm[quarter_cols].isna().all(axis=1).sum())

print("\n---------------- ALL-QUARTERS-MISSING ROWS ----------------")
print(f"Rows where every quarter is missing: {all_quarters_missing:,}")

# ------------------------------------------------------------
# 9. Fully complete rows
# ------------------------------------------------------------
all_quarters_present = int(qm[quarter_cols].notna().all(axis=1).sum())

print("\n---------------- FULLY COMPLETE ROWS ----------------")
print(f"Rows where every quarter is present: {all_quarters_present:,}")

# ------------------------------------------------------------
# 10. Long-form completeness view
# ------------------------------------------------------------
qm_long = qm.melt(
    id_vars=id_cols,
    value_vars=quarter_cols,
    var_name="quarter",
    value_name="score"
)

long_summary = pd.DataFrame([{
    "total_cells": len(qm_long),
    "nonmissing_cells": int(qm_long["score"].notna().sum()),
    "missing_cells": int(qm_long["score"].isna().sum()),
    "pct_nonmissing": qm_long["score"].notna().mean() * 100,
    "pct_missing": qm_long["score"].isna().mean() * 100
}])

print("\n---------------- OVERALL CELL COMPLETENESS ----------------")
print(long_summary.to_string(index=False))

# ------------------------------------------------------------
# 11. Sparse metrics
# ------------------------------------------------------------
sparse_metrics = metric_completeness.loc[
    metric_completeness["pct_cells_nonmissing"] < 25
].copy()

print("\n---------------- SPARSE METRICS (<25% NONMISSING) ----------------")
if len(sparse_metrics) == 0:
    print("None")
else:
    print(sparse_metrics.to_string(index=False))

# ------------------------------------------------------------
# 12. Sparse facilities
# ------------------------------------------------------------
sparse_facilities = facility_completeness.loc[
    facility_completeness["pct_cells_nonmissing"] < 25
].copy()

print("\n---------------- SPARSE FACILITIES (<25% NONMISSING) ----------------")
if len(sparse_facilities) == 0:
    print("None")
else:
    print(sparse_facilities.head(100).to_string(index=False))

print("\nDone. Files written to:")
print(out_dir / "quality_measures_missing_by_column.csv")
print(out_dir / "quality_measures_code_description_conflicts_from_panel.csv")
print(out_dir / "quality_measures_quarter_completeness.csv")
print(out_dir / "quality_measures_metric_completeness.csv")
print(out_dir / "quality_measures_facility_completeness.csv")
```

```python
from pathlib import Path
import pandas as pd
import numpy as np
import re

# ------------------------------------------------------------
# Paths
# ------------------------------------------------------------
PROJECT_ROOT = Path(r"C:/Repositories/white-bowblis-nhmc")
RAW_DIR = Path(r"C:\Users\Owner\OneDrive\NursingHomeData\quality-measures")
OUT_DIR = PROJECT_ROOT / "data" / "interim"
OUT_DIR.mkdir(parents=True, exist_ok=True)

# ------------------------------------------------------------
# Helpers
# ------------------------------------------------------------
def read_csv_with_fallbacks(path: Path, usecols=None) -> pd.DataFrame:
    attempts = [
        {"encoding": "utf-8", "low_memory": False, "usecols": usecols},
        {"encoding": "utf-8-sig", "low_memory": False, "usecols": usecols},
        {"encoding": "latin1", "low_memory": False, "usecols": usecols},
    ]
    last_err = None
    for kwargs in attempts:
        try:
            return pd.read_csv(path, **kwargs)
        except Exception as e:
            last_err = e
    raise last_err


def first_existing(cols, candidates):
    for c in candidates:
        if c in cols:
            return c
    return None


def normalize_ccn(x):
    if pd.isna(x):
        return pd.NA
    s = str(x).strip()
    s = re.sub(r"\.0$", "", s)
    s = re.sub(r"\D", "", s)
    if s == "":
        return pd.NA
    return s.zfill(6)


def parse_file_yyyy_mm(path: Path):
    m = re.search(r"quality_measures_(\d{4})_(\d{2})\.csv$", path.name)
    if not m:
        raise ValueError(f"Could not parse yyyy/mm from file name: {path.name}")
    return int(m.group(1)), int(m.group(2))


def pct_nonmissing(arr):
    return float(np.mean(~pd.isna(arr)) * 100)


# ------------------------------------------------------------
# Candidate schema names
# ------------------------------------------------------------
CCN_CANDS = ["CMS Certification Number (CCN)", "Federal Provider Number", "PROVNUM"]
CODE_CANDS = ["Measure Code", "MSR_CD"]
DESC_CANDS = ["Measure Description", "MSR_DESCR"]
PERIOD_CANDS = ["Measure Period", "MEASURE_PERIOD"]

Q_SCORE_CANDS = {
    1: ["Q1 Measure Score", "Q1_MEASURE_SCORE"],
    2: ["Q2 Measure Score", "Q2_MEASURE_SCORE"],
    3: ["Q3 Measure Score", "Q3_MEASURE_SCORE"],
    4: ["Q4 Measure Score", "Q4_MEASURE_SCORE"],
}

Q_FN_CANDS = {
    1: ["Footnote for Q1 Measure Score", "Q1_MEASURE_FN"],
    2: ["Footnote for Q2 Measure Score", "Q2_MEASURE_FN"],
    3: ["Footnote for Q3 Measure Score", "Q3_MEASURE_FN"],
    4: ["Footnote for Q4 Measure Score", "Q4_MEASURE_FN"],
}

AVG_SCORE_CANDS = ["Four Quarter Average Score", "MEASURE_SCORE_4QTR_AVG"]
AVG_FN_CANDS = ["Footnote for Four Quarter Average Score", "MEASURE_SCORE_4QTR_AVG_FN"]

# ------------------------------------------------------------
# Find files
# ------------------------------------------------------------
files = sorted(RAW_DIR.glob("quality_measures_*.csv"))
if not files:
    raise FileNotFoundError(f"No quality measure files found in {RAW_DIR}")

print(f"Found {len(files)} files.")

# ------------------------------------------------------------
# Containers
# ------------------------------------------------------------
file_summary_rows = []
column_missing_rows = []
metric_file_rows = []
facility_file_rows = []
code_desc_rows = []

# ------------------------------------------------------------
# Main loop
# ------------------------------------------------------------
for f in files:
    yyyy, mm = parse_file_yyyy_mm(f)
    file_id = f"{yyyy:04d}_{mm:02d}"
    print(f"Reading {f.name} ...")

    try:
        df = read_csv_with_fallbacks(f)
        cols = df.columns.tolist()

        ccn_col = first_existing(cols, CCN_CANDS)
        code_col = first_existing(cols, CODE_CANDS)
        desc_col = first_existing(cols, DESC_CANDS)
        period_col = first_existing(cols, PERIOD_CANDS)

        q_score_cols = {slot: first_existing(cols, Q_SCORE_CANDS[slot]) for slot in range(1, 5)}
        q_fn_cols = {slot: first_existing(cols, Q_FN_CANDS[slot]) for slot in range(1, 5)}
        avg_score_col = first_existing(cols, AVG_SCORE_CANDS)
        avg_fn_col = first_existing(cols, AVG_FN_CANDS)

        # ---------- File-level summary ----------
        row = {
            "file_name": f.name,
            "file_yyyy": yyyy,
            "file_mm": mm,
            "n_rows": len(df),
            "n_cols": len(df.columns),
            "ccn_col": ccn_col,
            "code_col": code_col,
            "desc_col": desc_col,
            "period_col": period_col,
            "q1_score_col": q_score_cols[1],
            "q2_score_col": q_score_cols[2],
            "q3_score_col": q_score_cols[3],
            "q4_score_col": q_score_cols[4],
            "avg_score_col": avg_score_col,
            "status": "ok"
        }

        # ---------- Missingness by column ----------
        for c in df.columns:
            column_missing_rows.append({
                "file_name": f.name,
                "file_yyyy": yyyy,
                "file_mm": mm,
                "column_name": c,
                "n_missing": int(df[c].isna().sum()),
                "pct_missing": float(df[c].isna().mean() * 100),
                "n_nonmissing": int(df[c].notna().sum()),
                "pct_nonmissing": float(df[c].notna().mean() * 100)
            })

        # ---------- Standardize key columns where present ----------
        work = df.copy()

        if ccn_col is not None:
            work["cms_certification_number"] = work[ccn_col].map(normalize_ccn)

        if code_col is not None:
            work["quality_metric_code"] = work[code_col].astype(str).str.strip()

        if desc_col is not None:
            work["quality_metric_description"] = work[desc_col].astype(str).str.strip()

        if period_col is not None:
            work["measure_period"] = work[period_col].astype(str).str.strip()

        for slot in range(1, 5):
            if q_score_cols[slot] is not None:
                work[f"q{slot}_score"] = pd.to_numeric(work[q_score_cols[slot]], errors="coerce")
            if q_fn_cols[slot] is not None:
                work[f"q{slot}_fn"] = work[q_fn_cols[slot]]

        if avg_score_col is not None:
            work["avg_score"] = pd.to_numeric(work[avg_score_col], errors="coerce")

        if avg_fn_col is not None:
            work["avg_fn"] = work[avg_fn_col]

        # ---------- File-level completeness for score slots ----------
        score_cols_present = [f"q{s}_score" for s in range(1, 5) if f"q{s}_score" in work.columns]
        if "avg_score" in work.columns:
            score_cols_present = score_cols_present + ["avg_score"]

        if score_cols_present:
            row["pct_cells_nonmissing_scores"] = pct_nonmissing(work[score_cols_present].to_numpy())
            row["n_rows_all_scores_missing"] = int(work[score_cols_present].isna().all(axis=1).sum())
            row["n_rows_any_score_present"] = int(work[score_cols_present].notna().any(axis=1).sum())
        else:
            row["pct_cells_nonmissing_scores"] = np.nan
            row["n_rows_all_scores_missing"] = np.nan
            row["n_rows_any_score_present"] = np.nan

        if all(f"q{s}_score" in work.columns for s in range(1, 5)):
            row["pct_q1_nonmissing"] = float(work["q1_score"].notna().mean() * 100)
            row["pct_q2_nonmissing"] = float(work["q2_score"].notna().mean() * 100)
            row["pct_q3_nonmissing"] = float(work["q3_score"].notna().mean() * 100)
            row["pct_q4_nonmissing"] = float(work["q4_score"].notna().mean() * 100)
        else:
            row["pct_q1_nonmissing"] = np.nan
            row["pct_q2_nonmissing"] = np.nan
            row["pct_q3_nonmissing"] = np.nan
            row["pct_q4_nonmissing"] = np.nan

        # ---------- Duplicate checks ----------
        if {"cms_certification_number", "quality_metric_code", "quality_metric_description"}.issubset(work.columns):
            id_df = work[["cms_certification_number", "quality_metric_code", "quality_metric_description"]].copy()
            row["n_duplicate_ccn_code_desc"] = int(id_df.duplicated().sum())
        else:
            row["n_duplicate_ccn_code_desc"] = np.nan

        if {"cms_certification_number", "quality_metric_code"}.issubset(work.columns):
            id2_df = work[["cms_certification_number", "quality_metric_code"]].copy()
            row["n_duplicate_ccn_code"] = int(id2_df.duplicated().sum())
        else:
            row["n_duplicate_ccn_code"] = np.nan

        # ---------- Code-description mapping by file ----------
        if {"quality_metric_code", "quality_metric_description"}.issubset(work.columns):
            pairs = (
                work[["quality_metric_code", "quality_metric_description"]]
                .dropna()
                .drop_duplicates()
                .copy()
            )
            pairs["file_name"] = f.name
            pairs["file_yyyy"] = yyyy
            pairs["file_mm"] = mm
            code_desc_rows.append(pairs)

        # ---------- Metric-level completeness within each file ----------
        if {"quality_metric_code", "quality_metric_description"}.issubset(work.columns):
            gcols = ["quality_metric_code", "quality_metric_description"]

            metric_grp = work.groupby(gcols, dropna=False)

            for (code, desc), g in metric_grp:
                out = {
                    "file_name": f.name,
                    "file_yyyy": yyyy,
                    "file_mm": mm,
                    "quality_metric_code": code,
                    "quality_metric_description": desc,
                    "n_rows": len(g),
                    "n_unique_ccns": g["cms_certification_number"].nunique(dropna=True) if "cms_certification_number" in g.columns else np.nan,
                }

                for slot in range(1, 5):
                    col = f"q{slot}_score"
                    out[f"pct_q{slot}_nonmissing"] = float(g[col].notna().mean() * 100) if col in g.columns else np.nan

                out["pct_avg_nonmissing"] = float(g["avg_score"].notna().mean() * 100) if "avg_score" in g.columns else np.nan

                present_score_cols = [c for c in [f"q1_score", f"q2_score", f"q3_score", f"q4_score", "avg_score"] if c in g.columns]
                if present_score_cols:
                    out["pct_cells_nonmissing_scores"] = pct_nonmissing(g[present_score_cols].to_numpy())
                else:
                    out["pct_cells_nonmissing_scores"] = np.nan

                metric_file_rows.append(out)

        # ---------- Facility-level completeness within each file ----------
        if "cms_certification_number" in work.columns:
            fac_grp = work.groupby("cms_certification_number", dropna=False)

            for ccn, g in fac_grp:
                out = {
                    "file_name": f.name,
                    "file_yyyy": yyyy,
                    "file_mm": mm,
                    "cms_certification_number": ccn,
                    "n_rows": len(g),
                    "n_unique_metrics": g["quality_metric_code"].nunique(dropna=True) if "quality_metric_code" in g.columns else np.nan,
                }

                for slot in range(1, 5):
                    col = f"q{slot}_score"
                    out[f"pct_q{slot}_nonmissing"] = float(g[col].notna().mean() * 100) if col in g.columns else np.nan

                out["pct_avg_nonmissing"] = float(g["avg_score"].notna().mean() * 100) if "avg_score" in g.columns else np.nan

                present_score_cols = [c for c in [f"q1_score", f"q2_score", f"q3_score", f"q4_score", "avg_score"] if c in g.columns]
                if present_score_cols:
                    out["pct_cells_nonmissing_scores"] = pct_nonmissing(g[present_score_cols].to_numpy())
                else:
                    out["pct_cells_nonmissing_scores"] = np.nan

                facility_file_rows.append(out)

        file_summary_rows.append(row)

    except Exception as e:
        file_summary_rows.append({
            "file_name": f.name,
            "file_yyyy": yyyy,
            "file_mm": mm,
            "n_rows": np.nan,
            "n_cols": np.nan,
            "status": f"ERROR: {e}"
        })

# ------------------------------------------------------------
# Build outputs
# ------------------------------------------------------------
file_summary = pd.DataFrame(file_summary_rows).sort_values(["file_yyyy", "file_mm"])
column_missing = pd.DataFrame(column_missing_rows).sort_values(["file_yyyy", "file_mm", "column_name"])
metric_file = pd.DataFrame(metric_file_rows).sort_values(["file_yyyy", "file_mm", "quality_metric_code"])
facility_file = pd.DataFrame(facility_file_rows).sort_values(["file_yyyy", "file_mm", "cms_certification_number"])

if code_desc_rows:
    code_desc_file = (
        pd.concat(code_desc_rows, ignore_index=True)
        .drop_duplicates()
        .sort_values(["quality_metric_code", "quality_metric_description", "file_yyyy", "file_mm"])
    )

    code_desc_conflicts_by_file = (
        code_desc_file[["file_name", "file_yyyy", "file_mm", "quality_metric_code", "quality_metric_description"]]
        .drop_duplicates()
        .groupby(["file_name", "file_yyyy", "file_mm", "quality_metric_code"], dropna=False)
        .size()
        .reset_index(name="n_descriptions")
    )
    code_desc_conflicts_by_file = (
        code_desc_conflicts_by_file
        .loc[code_desc_conflicts_by_file["n_descriptions"] > 1]
        .sort_values(["file_yyyy", "file_mm", "quality_metric_code"])
    )

    code_desc_conflicts_overall = (
        code_desc_file[["quality_metric_code", "quality_metric_description"]]
        .drop_duplicates()
        .groupby("quality_metric_code", dropna=False)
        .size()
        .reset_index(name="n_descriptions")
    )
    code_desc_conflicts_overall = (
        code_desc_conflicts_overall
        .loc[code_desc_conflicts_overall["n_descriptions"] > 1]
        .sort_values("quality_metric_code")
    )
else:
    code_desc_file = pd.DataFrame()
    code_desc_conflicts_by_file = pd.DataFrame()
    code_desc_conflicts_overall = pd.DataFrame()

# ------------------------------------------------------------
# Console summary
# ------------------------------------------------------------
print("\n---------------- BASIC SUMMARY ----------------")
print(f"Files processed: {len(file_summary):,}")
print(f"Successful files: {(file_summary['status'] == 'ok').sum():,}")
print(f"Failed files: {(file_summary['status'] != 'ok').sum():,}")

if not file_summary.empty:
    print("\nFirst few file summaries:")
    print(file_summary.head(10).to_string(index=False))

if not column_missing.empty:
    print("\nTop 25 most-missing file-columns:")
    print(column_missing.sort_values(["pct_missing", "n_missing"], ascending=[False, False]).head(25).to_string(index=False))

if not metric_file.empty:
    print("\nTop 25 sparsest metric-file combos by score completeness:")
    print(metric_file.sort_values("pct_cells_nonmissing_scores", ascending=True).head(25).to_string(index=False))

if not facility_file.empty:
    print("\nTop 25 sparsest facility-file combos by score completeness:")
    print(facility_file.sort_values("pct_cells_nonmissing_scores", ascending=True).head(25).to_string(index=False))

if not code_desc_conflicts_overall.empty:
    print("\nCodes with multiple descriptions across files:")
    print(code_desc_conflicts_overall.to_string(index=False))
else:
    print("\nNo cross-file code-description conflicts found.")

# ------------------------------------------------------------
# Save outputs
# ------------------------------------------------------------
file_summary.to_csv(OUT_DIR / "quality_measures_raw_file_summary.csv", index=False)
column_missing.to_csv(OUT_DIR / "quality_measures_raw_missing_by_file_column.csv", index=False)
metric_file.to_csv(OUT_DIR / "quality_measures_raw_metric_file_completeness.csv", index=False)
facility_file.to_csv(OUT_DIR / "quality_measures_raw_facility_file_completeness.csv", index=False)
code_desc_file.to_csv(OUT_DIR / "quality_measures_raw_code_description_by_file.csv", index=False)
code_desc_conflicts_by_file.to_csv(OUT_DIR / "quality_measures_raw_code_description_conflicts_by_file.csv", index=False)
code_desc_conflicts_overall.to_csv(OUT_DIR / "quality_measures_raw_code_description_conflicts_overall.csv", index=False)

print("\nDone. Files written to:")
print(OUT_DIR / "quality_measures_raw_file_summary.csv")
print(OUT_DIR / "quality_measures_raw_missing_by_file_column.csv")
print(OUT_DIR / "quality_measures_raw_metric_file_completeness.csv")
print(OUT_DIR / "quality_measures_raw_facility_file_completeness.csv")
print(OUT_DIR / "quality_measures_raw_code_description_by_file.csv")
print(OUT_DIR / "quality_measures_raw_code_description_conflicts_by_file.csv")
print(OUT_DIR / "quality_measures_raw_code_description_conflicts_overall.csv")
```
