```python
#!/usr/bin/env python
# coding: utf-8

from __future__ import annotations

import pandas as pd
import config as cfg

INTERIM = cfg.INTERIM_DIR
OUT = cfg.ensure_dir(cfg.OUTPUTS_DIR / "tables")

CHOW_FP = INTERIM / "chow.csv"

OUT_HCRIS0_NHC1 = OUT / "ccns_hcris0_nhc1.csv"
OUT_HCRIS1_NHC0 = OUT / "ccns_hcris1_nhc0.csv"

chow = pd.read_csv(CHOW_FP, low_memory=False)

chow["cms_certification_number"] = cfg.normalize_ccn_any(
    chow["cms_certification_number"]
)

chow["n_chow_nh_compare"] = (
    pd.to_numeric(chow["n_chow_nh_compare"], errors="coerce")
    .fillna(0)
    .astype(int)
)

chow["n_chow_mcr"] = (
    pd.to_numeric(chow["n_chow_mcr"], errors="coerce")
    .fillna(0)
    .astype(int)
)

# Bucket 1: 0 in HCRIS/MCR, 1 in NHC
hcris0_nhc1 = (
    chow.loc[
        (chow["n_chow_mcr"] == 0) &
        (chow["n_chow_nh_compare"] == 1),
        ["cms_certification_number"]
    ]
    .drop_duplicates()
    .sort_values("cms_certification_number")
    .reset_index(drop=True)
)

# Bucket 2: 1 in HCRIS/MCR, 0 in NHC
hcris1_nhc0 = (
    chow.loc[
        (chow["n_chow_mcr"] == 1) &
        (chow["n_chow_nh_compare"] == 0),
        ["cms_certification_number"]
    ]
    .drop_duplicates()
    .sort_values("cms_certification_number")
    .reset_index(drop=True)
)

hcris0_nhc1.to_csv(OUT_HCRIS0_NHC1, index=False)
hcris1_nhc0.to_csv(OUT_HCRIS1_NHC0, index=False)

print(f"HCRIS = 0, NHC = 1: {len(hcris0_nhc1):,} unique CCNs")
print(f"HCRIS = 1, NHC = 0: {len(hcris1_nhc0):,} unique CCNs")

print("\nSaved:")
print(OUT_HCRIS0_NHC1)
print(OUT_HCRIS1_NHC0)
```

```python
#!/usr/bin/env python
# coding: utf-8

from __future__ import annotations

import pandas as pd
import config as cfg

INTERIM = cfg.INTERIM_DIR
OUT = cfg.ensure_dir(cfg.OUTPUTS_DIR / "tables")

# Inputs
SIG_FP = INTERIM / "facility_signatures_long.csv"

HCRIS0_NHC1_FP = OUT / "ccns_hcris0_nhc1.csv"
HCRIS1_NHC0_FP = OUT / "ccns_hcris1_nhc0.csv"

# Outputs
OUT_SIG_HCRIS0_NHC1 = OUT / "facility_signatures_hcris0_nhc1.csv"
OUT_SIG_HCRIS1_NHC0 = OUT / "facility_signatures_hcris1_nhc0.csv"


def read_ccn_list(fp):
    df = pd.read_csv(fp, low_memory=False)

    df["cms_certification_number"] = cfg.normalize_ccn_any(
        df["cms_certification_number"]
    )

    return (
        df[["cms_certification_number"]]
        .dropna()
        .drop_duplicates()
    )


def main():
    signatures = pd.read_csv(SIG_FP, low_memory=False)

    signatures["cms_certification_number"] = cfg.normalize_ccn_any(
        signatures["cms_certification_number"]
    )

    hcris0_nhc1 = read_ccn_list(HCRIS0_NHC1_FP)
    hcris1_nhc0 = read_ccn_list(HCRIS1_NHC0_FP)

    # Bucket 1: 0 in HCRIS/MCR, 1 in NHC
    sig_hcris0_nhc1 = signatures.merge(
        hcris0_nhc1,
        on="cms_certification_number",
        how="inner"
    )

    # Bucket 2: 1 in HCRIS/MCR, 0 in NHC
    sig_hcris1_nhc0 = signatures.merge(
        hcris1_nhc0,
        on="cms_certification_number",
        how="inner"
    )

    sig_hcris0_nhc1 = sig_hcris0_nhc1.sort_values(
        ["cms_certification_number", "group_n"]
    ).reset_index(drop=True)

    sig_hcris1_nhc0 = sig_hcris1_nhc0.sort_values(
        ["cms_certification_number", "group_n"]
    ).reset_index(drop=True)

    sig_hcris0_nhc1.to_csv(OUT_SIG_HCRIS0_NHC1, index=False)
    sig_hcris1_nhc0.to_csv(OUT_SIG_HCRIS1_NHC0, index=False)

    print("[input counts]")
    print(f"HCRIS = 0, NHC = 1 CCNs: {hcris0_nhc1['cms_certification_number'].nunique():,}")
    print(f"HCRIS = 1, NHC = 0 CCNs: {hcris1_nhc0['cms_certification_number'].nunique():,}")

    print("\n[signature rows]")
    print(f"HCRIS = 0, NHC = 1 signature rows: {len(sig_hcris0_nhc1):,}")
    print(f"HCRIS = 1, NHC = 0 signature rows: {len(sig_hcris1_nhc0):,}")

    print("\n[signature CCNs matched]")
    print(f"HCRIS = 0, NHC = 1 matched CCNs: {sig_hcris0_nhc1['cms_certification_number'].nunique():,}")
    print(f"HCRIS = 1, NHC = 0 matched CCNs: {sig_hcris1_nhc0['cms_certification_number'].nunique():,}")

    print("\n[saved]")
    print(OUT_SIG_HCRIS0_NHC1)
    print(OUT_SIG_HCRIS1_NHC0)


if __name__ == "__main__":
    main()
```
