```python
import pandas as pd
import numpy as np

# -----------------------------------------------------------------------------
# Path
# -----------------------------------------------------------------------------
fp = r"C:\Repositories\white-bowblis-nhmc\data\clean\quality_panel.csv"

# -----------------------------------------------------------------------------
# Load
# -----------------------------------------------------------------------------
df = pd.read_csv(fp, low_memory=False)

print("=" * 80)
print("QUALITY PANEL QA REPORT")
print("=" * 80)
print(f"Rows: {len(df):,}")
print(f"Columns: {df.shape[1]:,}")

# -----------------------------------------------------------------------------
# Basic key checks
# -----------------------------------------------------------------------------
required = ["cms_certification_number", "year", "quarter"]
missing_required = [c for c in required if c not in df.columns]
if missing_required:
    print("\n[ERROR] Missing required key columns:", missing_required)
else:
    df["cms_certification_number"] = df["cms_certification_number"].astype(str).str.strip()
    df["year"] = pd.to_numeric(df["year"], errors="coerce")
    df["quarter"] = df["quarter"].astype(str).str.strip().str.upper()

    dup_mask = df.duplicated(subset=required, keep=False)
    n_dup_rows = int(dup_mask.sum())
    n_dup_keys = int(df.loc[dup_mask, required].drop_duplicates().shape[0])

    print("\nKEY CHECKS")
    print("-" * 80)
    print(f"Unique facilities: {df['cms_certification_number'].nunique():,}")
    print(f"Duplicate facility-year-quarter rows: {n_dup_rows:,}")
    print(f"Duplicate facility-year-quarter keys: {n_dup_keys:,}")

# -----------------------------------------------------------------------------
# Quarter coverage
# -----------------------------------------------------------------------------
if all(c in df.columns for c in ["year", "quarter"]):
    tmp = df.copy()
    qnum = tmp["quarter"].str.extract(r"Q([1-4])", expand=False)
    tmp["qnum"] = pd.to_numeric(qnum, errors="coerce")
    tmp["year_quarter"] = tmp["year"].astype("Int64").astype(str) + "Q" + tmp["qnum"].astype("Int64").astype(str)

    q_counts = (
        tmp.groupby("year_quarter", dropna=False)
        .size()
        .reset_index(name="n_rows")
        .sort_values(["year_quarter"])
    )

    print("\nQUARTER COVERAGE")
    print("-" * 80)
    print(q_counts.to_string(index=False))

# -----------------------------------------------------------------------------
# Missingness by column
# -----------------------------------------------------------------------------
miss = pd.DataFrame({
    "column": df.columns,
    "n_missing": df.isna().sum().values,
    "pct_missing": (df.isna().mean().values * 100)
}).sort_values(["pct_missing", "n_missing"], ascending=[False, False])

print("\nCOLUMN MISSINGNESS (top 40)")
print("-" * 80)
print(miss.head(40).to_string(index=False, formatters={"pct_missing": "{:.2f}".format}))

# -----------------------------------------------------------------------------
# Outcome coverage
# -----------------------------------------------------------------------------
qm_cols = sorted([c for c in df.columns if c.startswith("qm_")])

if qm_cols:
    qm_summary = pd.DataFrame({
        "outcome": qm_cols,
        "n_nonmissing": [df[c].notna().sum() for c in qm_cols],
        "pct_nonmissing": [df[c].notna().mean() * 100 for c in qm_cols],
        "mean": [pd.to_numeric(df[c], errors="coerce").mean() for c in qm_cols],
        "sd": [pd.to_numeric(df[c], errors="coerce").std() for c in qm_cols],
        "min": [pd.to_numeric(df[c], errors="coerce").min() for c in qm_cols],
        "max": [pd.to_numeric(df[c], errors="coerce").max() for c in qm_cols],
    }).sort_values("pct_nonmissing")

    print("\nQUALITY OUTCOME COVERAGE")
    print("-" * 80)
    print(qm_summary.to_string(index=False, formatters={
        "pct_nonmissing": "{:.2f}".format,
        "mean": "{:.4f}".format,
        "sd": "{:.4f}".format,
        "min": "{:.4f}".format,
        "max": "{:.4f}".format,
    }))

    if "year_quarter" not in df.columns and all(c in df.columns for c in ["year", "quarter"]):
        qnum = df["quarter"].astype(str).str.extract(r"Q([1-4])", expand=False)
        df["year_quarter"] = df["year"].astype("Int64").astype(str) + "Q" + pd.to_numeric(qnum, errors="coerce").astype("Int64").astype(str)

    if "year_quarter" in df.columns:
        print("\nQUALITY OUTCOME COVERAGE BY QUARTER")
        print("-" * 80)
        cov_by_q = (
            df.groupby("year_quarter")[qm_cols]
            .apply(lambda x: x.notna().mean() * 100)
            .reset_index()
        )
        print(cov_by_q.to_string(index=False, formatters={c: "{:.1f}".format for c in qm_cols}))
else:
    print("\nNo qm_ outcome columns found.")

# -----------------------------------------------------------------------------
# PBJ QA flags
# -----------------------------------------------------------------------------
pbj_flags = [
    "pbj_partial_quarter",
    "pbj_low_coverage",
    "pbj_zero_rn_lpn",
    "pbj_implausible_hprd",
    "pbj_invalid_quarter",
]

existing_pbj_flags = [c for c in pbj_flags if c in df.columns]
if existing_pbj_flags:
    print("\nPBJ QA FLAGS")
    print("-" * 80)
    for c in existing_pbj_flags:
        vals = pd.to_numeric(df[c], errors="coerce")
        n_flag = int((vals == 1).sum())
        pct_flag = float((vals == 1).mean() * 100)
        print(f"{c:25s} flagged={n_flag:>8,}   pct={pct_flag:>6.2f}%")

# -----------------------------------------------------------------------------
# Treatment/event-time sanity
# -----------------------------------------------------------------------------
sanity_cols = [c for c in ["treated", "post", "event_time", "time", "time_treated"] if c in df.columns]
if sanity_cols:
    print("\nTREATMENT / TIMING SANITY")
    print("-" * 80)

    for c in ["treated", "post"]:
        if c in df.columns:
            vals = pd.to_numeric(df[c], errors="coerce")
            print(f"{c:12s} value counts:")
            print(vals.value_counts(dropna=False).sort_index().to_string())
            print()

    if "event_time" in df.columns and "treated" in df.columns:
        treated = pd.to_numeric(df["treated"], errors="coerce")
        et = pd.to_numeric(df["event_time"], errors="coerce")

        print(f"Treated rows with missing event_time: {int(((treated == 1) & et.isna()).sum()):,}")
        print(f"Untreated rows with nonmissing event_time: {int(((treated == 0) & et.notna()).sum()):,}")

        if "post" in df.columns:
            post = pd.to_numeric(df["post"], errors="coerce")
            mismatch = ((post == 1) & (et <= 0)) | ((post == 0) & (et > 0) & et.notna())
            print(f"Post/event_time mismatches: {int(mismatch.sum()):,}")

# -----------------------------------------------------------------------------
# Key controls summary
# -----------------------------------------------------------------------------
control_cols = [
    "beds", "beds_prov", "num_beds",
    "occupancy_rate", "pct_medicare", "pct_medicaid",
    "rn_hprd", "lpn_hprd", "cna_hprd", "total_hprd",
    "coverage_ratio"
]
control_cols = [c for c in control_cols if c in df.columns]

if control_cols:
    ctrl = df[control_cols].apply(pd.to_numeric, errors="coerce")
    ctrl_summary = ctrl.describe(percentiles=[0.01, 0.05, 0.50, 0.95, 0.99]).T

    print("\nKEY CONTROL SUMMARY")
    print("-" * 80)
    keep_stats = ["count", "mean", "std", "min", "1%", "5%", "50%", "95%", "99%", "max"]
    keep_stats = [c for c in keep_stats if c in ctrl_summary.columns]
    print(ctrl_summary[keep_stats].to_string(float_format=lambda x: f"{x:,.4f}"))

# -----------------------------------------------------------------------------
# Facility-quarter continuity
# -----------------------------------------------------------------------------
if all(c in df.columns for c in ["cms_certification_number", "year", "quarter"]):
    tmp = df[["cms_certification_number", "year", "quarter"]].copy()
    tmp["qnum"] = pd.to_numeric(tmp["quarter"].astype(str).str.extract(r"Q([1-4])", expand=False), errors="coerce")
    tmp = tmp.dropna(subset=["year", "qnum"])
    tmp["qi"] = tmp["year"].astype(int) * 4 + tmp["qnum"].astype(int)
    tmp = tmp.sort_values(["cms_certification_number", "qi"])

    tmp["gap_from_prev_quarters_calc"] = (
        tmp.groupby("cms_certification_number")["qi"]
        .diff()
        .fillna(1)
        .astype(int) - 1
    ).clip(lower=0)

    gap_counts = tmp["gap_from_prev_quarters_calc"].value_counts().sort_index()

    print("\nFACILITY-QUARTER CONTINUITY")
    print("-" * 80)
    print("Calculated gap_from_prev_quarters distribution:")
    print(gap_counts.to_string())

# -----------------------------------------------------------------------------
# Outcome + filter cross-tabs
# -----------------------------------------------------------------------------
if qm_cols and existing_pbj_flags:
    print("\nOUTCOME COVERAGE AMONG PBJ-FLAGGED ROWS")
    print("-" * 80)
    for flag in existing_pbj_flags:
        flagged = pd.to_numeric(df[flag], errors="coerce") == 1
        if flagged.sum() == 0:
            continue
        row = {"flag": flag, "n_flagged": int(flagged.sum())}
        for q in qm_cols:
            row[q] = round(df.loc[flagged, q].notna().mean() * 100, 1)
        print(pd.DataFrame([row]).to_string(index=False))

print("\nDone.")
```
